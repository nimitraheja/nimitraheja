Bivariate analysis

```python
import pandas as pd
import seaborn as sns
tips=pd.read_csv('/content/tip.csv')
titanic=pd.read_csv('/content/train (1).csv')
iris=pd.read_csv('/content/iris (1).csv')
```

```python
tips.head()
```

scatterplot-numerical to numerical

```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', style='smoker', data=tips)
```

```python
import seaborn as sns
sns.barplot(x='Pclass', y='Age', data=titanic,hue='Sex')
```

boxplot

```python
sns.boxplot(x='Sex',y='Age',data=titanic)
```

distplot

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False)
```

heatmap(categorical-categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Embarked')['Survived'].mean()*100).plot(kind='bar')
```

clustermap(categorical-categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```
